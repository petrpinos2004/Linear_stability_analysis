```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.linalg import eig


def create_chebyshev_grid(N):
  """Generates Chebyshev-Lobatto nodes on [-1, 1]."""
  j = np.arange(N + 1)
  z = np.cos(np.pi * j / N)
  return z


def get_derivatives(N):
  """Computes 1st and 2nd Chebyshev differentiation matrices."""
  z = create_chebyshev_grid(N)
  j = np.arange(N + 1)

  c = np.ones(N + 1)
  c[0] = 2.0
  c[N] = 2.0
  c = c * (-1) ** j

  X = np.outer(z, np.ones(N + 1))
  dX = X - X.T  # grid points - transpose grid points to include delta_x_ij
  D1 = np.outer(c, 1.0 / c) / (dX + np.eye(N + 1))
  D1 = D1 - np.diag(np.sum(D1.T, axis=0))

  D2 = np.dot(D1, D1)
  return z, D1, D2


def build_system_matrix(N, Pr, Ra_th, kx, D_val, alpha_b, drho_dT, drho_dp, cp_b):
  """Assembles the 4-variable system: [u_x, u_z, p, T]

  corresponding to equations (5.5) - (5.9)[cite: 1, 2].
  """
  z, D1, D2 = get_derivatives(N)
  n_pts = N + 1
  dim = 4 * n_pts  # Fields: u_x (0), u_z (1), p (2), T (3)

  A = np.zeros((dim, dim))
  B = np.zeros((dim, dim))

  I = np.eye(n_pts)

  # Placeholder background states (can be modified to vary with z)
  rho_b = np.ones_like(z)
  drho_dz = np.zeros_like(z)
  Tb = 1.0 - 0.5 * z
  dTb_dz = -0.5 * np.ones_like(z)
  dp_dz = -rho_b  # Hydrostatic balance example

  # --- Block Slices ---
  ux_slice = slice(0, n_pts)
  uz_slice = slice(n_pts, 2 * n_pts)
  p_slice = slice(2 * n_pts, 3 * n_pts)
  T_slice = slice(3 * n_pts, 4 * n_pts)

  # --- Row 1: Eq. (5.5) + (5.9) [Continuity / Density disturbance] ---
  B[ux_slice, p_slice] = drho_dp * I
  B[ux_slice, T_slice] = drho_dT * I

  A[ux_slice, ux_slice] = -1j * kx * np.diag(rho_b)
  A[ux_slice, uz_slice] = -np.diag(rho_b) @ D1 - np.diag(drho_dz)

  # --- Row 2: Eq. (5.6) [u_x Momentum] ---
  B[uz_slice, ux_slice] = Pr ** (-1) * np.diag(rho_b)

  A[uz_slice, ux_slice] = D2 - (4.0 / 3.0) * (kx**2) * I
  A[uz_slice, uz_slice] = (1j * kx / 3.0) * D1
  A[uz_slice, p_slice] = -Ra_th * 1j * kx * I

  # --- Row 3: Eq. (5.7) + (5.9) [u_z Momentum] ---
  B[p_slice, uz_slice] = Pr ** (-1) * np.diag(rho_b)

  A[p_slice, ux_slice] = (1j * kx / 3.0) * D1
  A[p_slice, uz_slice] = (4.0 / 3.0) * D2 - (kx**2) * I
  A[p_slice, p_slice] = -Ra_th * D1 - Ra_th * drho_dp * I
  A[p_slice, T_slice] = -Ra_th * drho_dT * I

  # --- Row 4: Eq. (5.8) [Energy Equation] ---
  B[T_slice, p_slice] = -D_val * alpha_b * np.diag(Tb)
  B[T_slice, T_slice] = np.diag(rho_b * cp_b)

  A[T_slice, uz_slice] = np.diag(
      -rho_b * cp_b * dTb_dz + D_val * alpha_b * Tb * dp_dz
  )
  A[T_slice, T_slice] = D2 - (kx**2) * I

  # --- Boundary Conditions (Dirichlet: u_x=0, u_z=0, T=0 at walls; p is free) ---
  for b_idx in [0, N]:
    for field_base in [0, n_pts, 3 * n_pts]:  # u_x, u_z, T boundaries
      row = b_idx + field_base
      A[row, :] = 0
      B[row, :] = 0
      A[row, row] = 1.0

  return A, B


def solve_stability(A, B):
  """Solves the generalized eigenvalue problem and filters out spurious modes."""
  eigenvalues, _ = eig(A, B)

  # Filter out infinite or extreme numerical artifacts
  valid = np.isfinite(eigenvalues) & (np.abs(eigenvalues) < 1e5)
  filtered_eig = eigenvalues[valid]

  return filtered_eig


def plot_spectrum(eigenvalues):
  """Plots the complex eigenvalue spectrum (Real vs Imaginary growth rates)."""
  plt.figure(figsize=(8, 6))

  real_parts = np.real(eigenvalues)
  imag_parts = np.imag(eigenvalues)

  # Separate stable vs unstable modes for visualization
  unstable = real_parts > 0
  stable = ~unstable

  plt.scatter(
      real_parts[stable],
      imag_parts[stable],
      color="blue",
      alpha=0.6,
      label=r"Stable",
  )
  plt.scatter(
      real_parts[unstable],
      imag_parts[unstable],
      color="red",
      alpha=0.8,
      s=40,
      label="Unstable ($\text{Re}(\sigma) > 0$)",
  )

  plt.axvline(x=0, color="black", linestyle="--", linewidth=1)
  plt.xlabel("Growth Rate ($\text{Re}(\sigma)$)", fontsize=12)
  plt.ylabel("Oscillation Frequency ($\text{Im}(\sigma)$)", fontsize=12)
  plt.title("Linear Stability Eigenvalue Spectrum", fontsize=14)
  plt.grid(True, linestyle=":", alpha=0.7)
  plt.legend(loc="lower left")
  plt.tight_layout()
  plt.show()


def main():
  # Parameters
  N = 40
  Pr = 1.0
  Ra_th = 1e3
  kx = 3.117
  D_val = 0.1
  alpha_b = 1.0
  drho_dT = -1.0  # Thermal expansion effect on density
  drho_dp = 1.0  # Compressibility effect on density
  cp_b = 1.0

  # Build matrices and solve
  A, B = build_system_matrix(
      N, Pr, Ra_th, kx, D_val, alpha_b, drho_dT, drho_dp, cp_b
  )
  eigenvalues = solve_stability(A, B)

  # Sort by real part descending for terminal reporting
  sorted_sigma = eigenvalues[np.argsort(np.real(eigenvalues))[::-1]]
  max_sigma = np.real(sorted_sigma[0]) if len(sorted_sigma) > 0 else 0.0

  print(f"\n--- Results ---")
  print(f"Max Growth Rate (sigma): {max_sigma:.5f}")
  if max_sigma > 0:
    print("Status: UNSTABLE")
  else:
    print("Status: STABLE")

  # Generate the plot
  plot_spectrum(eigenvalues)


if __name__ == "__main__":
  main()
```
